```python
import pandas as pd

# Load data
orders = pd.read_excel(
    "Customer_Sales_Intelligence_cleaned.xlsx",
    sheet_name="Customer_Sales_Intelligence_cle"
)

# Analysis date
analysis_date = orders['Order_Date'].max()

# Recency
recency = orders.groupby('Customer_ID').agg(
    Last_Order_Date=('Order_Date', 'max')
).reset_index()

recency['Recency_Days'] = (
    analysis_date - recency['Last_Order_Date']
).dt.days

# Frequency
frequency = orders.groupby('Customer_ID').agg(
    Frequency=('Order_ID', 'nunique')
).reset_index()

# Monetary
monetary = orders.groupby('Customer_ID').agg(
    Monetary=('Sales', 'sum')
).reset_index()

# Build RFM table
rfm = (
    recency
    .merge(frequency, on='Customer_ID')
    .merge(monetary, on='Customer_ID')
)

# RFM Scores
rfm['R_Score'] = pd.qcut(
    rfm['Recency_Days'],
    q=5,
    labels=[5, 4, 3, 2, 1]
)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    q=5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    q=5,
    labels=[1, 2, 3, 4, 5]
)

# RFM Score
rfm['RFM_Score'] = (
    rfm['R_Score'].astype(int).astype(str) +
    rfm['F_Score'].astype(int).astype(str) +
    rfm['M_Score'].astype(int).astype(str)
)

# Customer Segment
def rfm_segment(row):
    r = row['R_Score']
    f = row['F_Score']
    m = row['M_Score']

    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'
    elif r >= 3 and f >= 3:
        return 'Loyal Customers'
    elif r >= 4 and f <= 2:
        return 'New Customers'
    elif r <= 2 and f >= 3:
        return 'At Risk'
    elif r <= 2 and f <= 2:
        return 'Lost Customers'
    else:
        return 'Potential Customers'

rfm['Customer_Segment'] = rfm.apply(rfm_segment, axis=1)

# Customer Value Segment
rfm['Value_Segment'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    q=3,
    labels=['Low Value', 'Medium Value', 'High Value']
)

# Save updated CSV
rfm.to_csv("RFM_Customer_Segments.csv", index=False)

# Check
print(rfm.columns.tolist())
```
